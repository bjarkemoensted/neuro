### 1.1.1 Trigonometric functions

```python
def make_trig_func(
        trig: Callable[[float], float]=np.sin,
        A: float=1.0,
        omega: float=1.0
        ) -> Callable[[float], float]:
    """Generates a trigonometric function like A*sin(omega*t) - trig is sin/cos/etc"""

    def y(t: float) -> float:
        res = A*trig(omega*t)
        return res
    return y
```

```python
y1 = make_trig_func(np.sin)
y2 = make_trig_func(np.cos)
```

```python
ax = plot_function(y1, xmax=10, label="$y_1$")
plot_function(y2, xmax=10, ax=ax, label="$y_2$")
ax.legend();
```

Try changing $\omega$

```python
ax = plt.subplot()
for omega in (0.5, 1.0, 2.0):
    plot_function(
        make_trig_func(omega=omega),
        ax=ax,
        xmax=20,
        label=f"$\\omega = {omega}$"
    )
ax.legend();
```

Heavens, $\omega$ is the frequency!

```python
ax = plt.subplot()
for A in (0.5, 1.0, 2.0):
    plot_function(
        make_trig_func(A=A),
        ax=ax,
        xmax=20,
        label=f"$A = {A}$"
    )
ax.legend();
```

And $A$ is the amplitude!!

### 1.1.2 Power functions

```python
n_vals = (2, 1, 0, -1)

ax = plt.subplot()
for n in n_vals:
    plot_function(lambda x: x**n, ax=ax, label=f"$x^{{{n}}}$", xmax=5, xmin=0.05, ylabel=None)
ax.legend();
```

Great Scott! $x^0$ is constant at 1. They're all $f(1) = 1$, $n<0$ diverges towards $x=0$.

```python
n_vals = (2, 1, 0, -1)

ax = plt.subplot()
for n in n_vals:
    plot_function(lambda x: x**n, ax=ax, label=f"$x^{{{n}}}$", xmax=5, xmin=0.05, ylabel=None)

ax.legend();
ax.loglog();
```

1.1.3 Exponential function

```python
def make_exponential(tau: float) -> Callable[[float], float]:
    def exp(t: float) -> float:
        return np.exp(t/tau)
    return exp
```

```python
taus = (-1, 1, 2, 3)
_, (ax1, ax2) = plt.subplots(nrows=2)
for tau in taus:
    label = f"$\\tau = {tau}$"
    kwargs = {"func": lambda t: np.exp(t/tau), "label": label, "xmin": -5, "xmax": 5}
    plot_function(ax=ax1, **kwargs)
    plot_function(ax=ax2, **kwargs)

ax1.legend();
ax2.semilogy();
```

*WAT* it's like $\tau$ is some time scale or something!
The lines! They're straight again!!

1.1.4 Gaussian function (normal distribution)

```python
def make_gaussian(mu: float=1.0, sigma: float=1.0) -> Callable[[float], float]:
    def gauss(x: float) -> float:
        return (1/sigma*np.sqrt(2*np.pi))*np.exp(-(mu - x)**2 / (2*sigma**2))
    return gauss
```

```python
pars = (
    (1.0, 1.0),
    (3.0, 2.0),
    (5.0, 0.2),
)

ax = plt.subplot()

for mu, sigma in pars:
    pars = dict(mu=mu, sigma=sigma)
    f = make_gaussian(**pars)
    label = plotting.formatting.label_with_values(pars)
    plot_function(f, ax=ax, xmin=-3, xmax=7, label=label)
ax.legend();
```

Pretty! Seems $\sigma$ spreads it out and $\mu$ shifts it!

```python
plot_function(make_gaussian()).loglog();
```

This kills the nice bell shape :/

1.1.5 Comparing Gaussian distribution with power laws

```python
def power(x: float) -> float:
    return x**-2

gaussian = make_gaussian()

ax = plot_function(gaussian, label="Gaussian")


plot_function(power, xmin=0.5, ax=ax, label="Power");
plot_function(lambda x: power(x)/gaussian(x), xmin=0.5, xmax=4.3, ax=ax, label="Ratio");
ax.legend();
```

It seems the Gaussian falls off the quickest - the ratio diverges (power-Gaussian ratio is a L'Hôpital where the power thingy dies after $n$ applications but the Gaussian is like impervious)

1.1.6 Sigmoidal functions

```python
a_vals = (0.4, 0.7, 1.0, 1.5)

ax = plt.subplot()

for a in reversed(a_vals):
    plot_function(
        lambda x: 1.0/(1 + np.exp(-a*x)),
        ax=ax,
        label=f"$a={a}$",
        xmin=-5,
        xmax=5
    )
ax.legend();
ax.set_ylabel(f"$\\frac{{1}}{{1+e^{{-ax}}}}$");
```

It's as if decreasing $a$ stretches the curve horizontally! (bc it's equivalent to rescaling $x$)
