### 1.1.1 Trigonometric functions

```python
def make_label(i: int, A: float, omega: float) -> str:
    return f"$y_{i}, A_{i}={A:.2g}, \\omega_{i}={omega:.2g}$"

def plot_trig_funcs(
        ax: Axes,
        A1: float=1,
        A2: float=1,
        omega1: float=1,
        omega2: float=1,
        t_max: float=20.0
        ) -> None:
    # TODO nah just have the function plot to an existing axis object!
    t = np.linspace(start=0.0, stop=t_max, num=10_000)
    y1 = A1*np.sin(omega1*t)
    y2 = A2*np.cos(omega2*t)

    ax.plot(t, y1, label=make_label(1, A1, omega1))
    ax.plot(t, y2, label=make_label(2, A2, omega2))

    ax.legend(loc="lower left")
```

```python
def make_trig_func(trig: Callable[[float], float], A: float=1.0, omega: float=1.0) -> Callable[[float], float]:
    def y(t: float) -> float:
        res = A*trig(omega*t)
        return res
    return y
```

```python
y1 = make_trig_func(np.sin)
y2 = make_trig_func(np.cos)
```

```python
ax = plotting.hacks.plot_function(y1, xmax=10)
plotting.hacks.plot_function(y2, xmax=10, ax=ax)
```
