### 1.1.1 Trigonometric functions

The cosine and sine functions are the classical oscillatory functions and often appear in dynamical systems. Plot sine and cosine:

$$
\begin{aligned}
y_1 &= A_1 \sin(\omega_1 t) \\
y_2 &= A_2 \cos(\omega_2 t)
\end{aligned}
\tag{1}
$$

$\omega$ is called the angular frequency, i.e. how quickly the angle changes over time as opposed to normal frequency, which is how many cycles the system goes through over time. They are related as:

$$
\omega = 2\pi f
$$

Try changing $\omega$. Try changing the amplitudes $A_1$ and $A_2$. Try plotting $y_1$ against $y_2$. What do you see? How does this look if $A_1$ is bigger than $A_2$?

```python
def make_trig_func(
        trig: Callable[[float], float]=np.sin,
        A: float=1.0,
        omega: float=1.0
        ) -> Callable[[float], float]:
    """Generates a trigonometric function like A*sin(omega*t) - trig is sin/cos/etc"""

    def y(t: float) -> float:
        res = A*trig(omega*t)
        return res
    return y
```

```python
y1 = make_trig_func(np.sin)
y2 = make_trig_func(np.cos)
```

```python
ax = plot_function(y1, xmax=10, label="$y_1$")
plot_function(y2, xmax=10, ax=ax, label="$y_2$")
ax.legend();
```

Try changing $\omega$

```python
ax = plt.subplot()
for omega in (0.5, 1.0, 2.0):
    plot_function(
        make_trig_func(omega=omega),
        ax=ax,
        xmax=20,
        label=f"$\\omega = {omega}$"
    )
ax.legend();
```

Heavens, $\omega$ is the frequency!

```python
ax = plt.subplot()
for A in (0.5, 1.0, 2.0):
    plot_function(
        make_trig_func(A=A),
        ax=ax,
        xmax=20,
        label=f"$A = {A}$"
    )
ax.legend();
```

### 1.1.4 Gaussian function (normal distribution)

The Gaussian function is also called the bell curve or normal distribution, since it is found to represent randomness in many physical and biological systems. The underlying cause of the Gaussian distribution is that random events are uncorrelated and add up linearly. The distribution looks as follows:

$$
f(x) = e^{-\frac{(x-\mu)^2}{2\sigma^2}}
\tag{4}
$$

This version does not include normalization. For simplicity, we skip the normalization here. $\mu$ represents the mean location (maximum) and $\sigma$ is the standard deviation.

Plot $f(x)$ for different values of $\mu$ and $\sigma$. Try plotting $f(x)$ on a double-log plot for $x>0$ and $\mu>0$. What happens to the nice bell shape?

```python
def make_gaussian(mu: float=1.0, sigma: float=1.0) -> Callable[[float], float]:
    def gauss(x: float) -> float:
        return (1/sigma*np.sqrt(2*np.pi))*np.exp(-(mu - x)**2 / (2*sigma**2))
    return gauss
```

```python
pars = (
    (1.0, 1.0),
    (3.0, 2.0),
    (5.0, 0.2),
)

ax = plt.subplot()

for mu, sigma in pars:
    pars = dict(mu=mu, sigma=sigma)
    f = make_gaussian(**pars)
    label = plotting.formatting.label_with_values(pars)
    plot_function(f, ax=ax, xmin=-3, xmax=7, label=label)
ax.legend();
```

Pretty! Seems $\sigma$ spreads it out and $\mu$ shifts it!

```python
plot_function(make_gaussian()).loglog();
```

This kills the nice bell shape :/

### 1.1.5 Comparing Gaussian distribution with power laws

Now you have plotted the Gaussian function on a double log. Take a moment to plot both the Gaussian and a power law function together on the same plot. Make sure the scaling exponent is negative.

In a race to the bottom as $x$ increases, which will win?

```python
def power(x: float) -> float:
    return x**-2

gaussian = make_gaussian()

ax = plot_function(gaussian, label="Gaussian")


plot_function(power, xmin=0.5, ax=ax, label="Power")
plot_function(lambda x: power(x)/gaussian(x), xmin=0.5, xmax=4.3, ax=ax, label="Ratio")
ax.legend();
```

It seems the Gaussian falls off the quickest - the ratio diverges (power-Gaussian ratio is a L'Hôpital where the power thingy dies after $n$ applications but the Gaussian is like impervious)

### 1.1.6 Sigmoidal functions

Why is the sigmoidal function fundamental? It is found in media with growth and saturation. It is used as a smooth threshold/activation function for simplified neuronal models. It can represent phase transitions and dose-response curves in pharmacology. It is also the most common neural network activation function in artificial networks.

The simplest form looks like this:

$$
\sigma(x) = \frac{1}{1+e^{-x}}
\tag{5}
$$

Other functions, like the hyperbolic tangent function, have a sigmoidal shape. Choice of function depends on circumstances.

Plot $\sigma$. Try putting a constant, $a>0$, in front of $x$ in the function and plot it with different values of $a$, starting with $a=1$. Plot these in the same coordinate system, for various values of $a$. What happens as $a$ becomes larger?

### 2.1.6 Eigenvalues and eigenvectors

What are the eigenvalues of:

$$
M =
\begin{bmatrix}
0 & 1 \\
1 & 0
\end{bmatrix}
$$

Finding the associated eigenvectors ($\mathbf{x}$) is done by inserting the eigenvalues into the equation and solving for $\mathbf{x}$:

$$
(M-\lambda I)\mathbf{x} =
\begin{bmatrix}
a-\lambda & b \\
c & d-\lambda
\end{bmatrix}
\begin{bmatrix}
x_1 \\
x_2
\end{bmatrix}
=
\begin{bmatrix}
0 \\
0
\end{bmatrix}
$$

Find the two eigenvectors.

Repeat the exercise for $M_2$: find eigenvalues and eigenvectors:

$$
M_2 =
\begin{bmatrix}
2 & -1 \\
-1 & 2
\end{bmatrix}
$$

Solution for $M_2$:

$$
\begin{aligned}
\lambda_1 &= 1, \qquad \lambda_2 = 3 \\
\mathbf{x}_1 &= 
\begin{bmatrix}
1 \\ 1
\end{bmatrix},
\qquad
\mathbf{x}_2 =
\begin{bmatrix}
1 \\ -1
\end{bmatrix}
\end{aligned}
$$

```python
M = np.array([[0, 1], [1, 0]], dtype=int)
M2 = np.array([[2, -1], [-1, 2]], dtype=int)
```

```python
def solve_eigen(M: npt.NDArray[np.floating]) -> None:
    evals, evecs = scipy.linalg.eig(M)
    # Convert eigenvalues from coplex to real after checking the imaginary part is zero
    assert np.all(evals == evals.real)
    evals = evals.real

    ax = plt.subplot()
    color_cycle = iter(matplotlib.rcParams['axes.prop_cycle'].by_key()['color'])

    print("Eigenvectors:")
    for i, x in enumerate(evecs.T):
        lambda_ = evals[i]
        x_scaled = lambda_*x

        print(f"*lambda {i}*")
        print(lambda_)
        print(f"*x_{i}*")
        # NOTE: Eigenvectors multiplied by any nonzero scalar are still eigenvectors, so length might vary
        print(x)
        print(f"*Mx_{i} = lambda_i * x_{i}?*")
        print(np.all(M.dot(x) == x_scaled))

        color = next(color_cycle)
        # Plot the eigenvector
        print("WTF", x)
        plotting.hacks.plot_2D_vector(
            x,
            color=color,
            ax=ax,
            label=f"$x_{i}$",
            quiver_kwargs={'alpha': 0.7}
        )

        # Plot the eigenvector after applying the operator
        plotting.hacks.plot_2D_vector(
                    M.dot(x),
                    color=color,
                    ax=ax,
                    label=f"$M\\cdot x_{i}$  ($\\lambda_{i} = {lambda_:.2g}$)",
                    quiver_kwargs={'alpha': 0.2}
                )

        print()
    ax.legend()
```

```python
solve_eigen(M)
```

Neat - the negative eigenvalue flips the vector around $y=x$, the positive one does nothing

```python
solve_eigen(M2)
```

The eigenvector with an eigenvalue of 3 stretches the vector by a factor of 3. The unit one does nothing.
